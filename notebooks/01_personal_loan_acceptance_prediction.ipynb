```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
personal_df = pd.read_csv("../data/processed/personal_loan_cleaned.csv")
approval_df = pd.read_csv("../data/processed/loan_approval_cleaned.csv")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=personal_df, x="Age", hue="Personal Loan", kde=True)
plt.title("Age Distribution by Personal Loan Acceptance")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=personal_df, x="Income", hue="Personal Loan", kde=True)
plt.title("Income Distribution by Personal Loan Acceptance")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=personal_df, x="Education", hue="Personal Loan")
plt.title("Education vs Personal Loan Acceptance")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(personal_df.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=approval_df, x="person_income", hue="loan_status", kde=True)
plt.title("Income Distribution by Loan Status")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=approval_df, x="loan_amnt", y="person_income", hue="loan_status")
plt.title("Loan Amount vs Income")
plt.show()
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

figures_dir = "../reports/figures"
os.makedirs(figures_dir, exist_ok=True)

graphs = [
    ("age_distribution_personal_loan.png", lambda: sns.histplot(data=personal_df, x="Age", hue="Personal Loan", kde=True), "Age Distribution by Personal Loan Acceptance"),
    ("income_distribution_personal_loan.png", lambda: sns.histplot(data=personal_df, x="Income", hue="Personal Loan", kde=True), "Income Distribution by Personal Loan Acceptance"),
    ("education_vs_personal_loan.png", lambda: sns.countplot(data=personal_df, x="Education", hue="Personal Loan"), "Education vs Personal Loan Acceptance"),
    ("personal_loan_correlation_heatmap.png", lambda: sns.heatmap(personal_df.corr(), annot=True, cmap="coolwarm", fmt=".2f"), "Personal Loan Correlation Heatmap"),
    ("income_distribution_loan_status.png", lambda: sns.histplot(data=approval_df, x="person_income", hue="loan_status", kde=True), "Income Distribution by Loan Status"),
    ("loan_amount_vs_income.png", lambda: sns.scatterplot(data=approval_df, x="loan_amnt", y="person_income", hue="loan_status"), "Loan Amount vs Income"),
]

for file_name, plot_func, title in graphs:
    plt.figure(figsize=(8, 5))
    plot_func()
    plt.title(title)
    plt.tight_layout()
    plt.savefig(os.path.join(figures_dir, file_name), dpi=300, bbox_inches="tight")
    plt.close()

print("All graphs saved successfully in reports/figures folder.")
```
